```python
import matplotlib.pyplot as plt
```

```python
init_fname = "../../../../../init_traj.txt"
```

```python
opt_fname = "../../../../../trajectory.txt"
```

# Only X Y in file

```python
# Load the INITIAL trajectory from the file
trajectory = []
with open(init_fname, "r") as f:
    for line in f:
        # Each line contains the X and Y positions
        x, y = map(float, line.strip().split())
        trajectory.append((x, y))

# Unzip the list of tuples into separate X and Y values
x_vals, y_vals = zip(*trajectory)

# Plot the trajectory
plt.figure(figsize=(8, 6))
plt.plot(x_vals, y_vals, marker='o', label="Initial Trajectory")
plt.xlabel('X Position')
plt.ylabel('Y Position')
plt.title('Initial Trajectory')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Load the trajectory from the file
trajectory = []
with open(opt_fname, "r") as f:
    for line in f:
        # Each line contains the X and Y positions
        x, y = map(float, line.strip().split())
        trajectory.append((x, y))

# Unzip the list of tuples into separate X and Y values
x_vals, y_vals = zip(*trajectory)

# Plot the trajectory
plt.figure(figsize=(8, 6))
plt.plot(x_vals, y_vals, marker='o', label="Optimized Trajectory")
plt.xlabel('X Position')
plt.ylabel('Y Position')
plt.title('Optimized Trajectory')
plt.legend()
plt.grid(True)
plt.show()
```

# Has full state and controls in file

```python

# Load the trajectory from the file
states = []
controls = []

with open(init_fname, "r") as f:
    for line in f:
        # Split the line into state (13 values) and control (6 values)
        data = list(map(float, line.strip().split()))
        state = data[:13]  # First 13 values are the state
        states.append(state)
        if len(data) > 13:
            control = data[13:]  # Remaining 6 values are the control
            controls.append(control)

# Extract X and Y positions from states
x_vals = [state[0] for state in states]  # X positions
y_vals = [state[1] for state in states]  # Y positions

# Plot the trajectory in the XY plane
plt.figure(figsize=(8, 6))
plt.plot(x_vals, y_vals, marker='o', label="Optimized Trajectory")
plt.xlabel('X Position')
plt.ylabel('Y Position')
plt.title('Optimized Trajectory in XY Plane')
plt.legend()
plt.grid(True)
plt.show()

# Optional: Plot controls if they exist
if controls:
    control_indices = range(len(controls))
    plt.figure(figsize=(8, 6))
    for i in range(6):
        control_vals = [control[i] for control in controls]
        plt.plot(control_indices, control_vals, label=f"u[{i}]")
    plt.xlabel('Time Step')
    plt.ylabel('Control Input')
    plt.title('Control Inputs over Time')
    plt.legend()
    plt.grid(True)
    plt.show()
```
