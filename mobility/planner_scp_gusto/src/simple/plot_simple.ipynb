```python
# Load the INITIAL trajectory from the file
trajectory = []
with open(init_fname, "r") as f:
    for line in f:
        # Each line contains the X and Y positions
        x, y = map(float, line.strip().split())
        trajectory.append((x, y))

# Unzip the list of tuples into separate X and Y values
x_vals, y_vals = zip(*trajectory)

# Plot the trajectory
plt.figure(figsize=(8, 6))
plt.plot(x_vals, y_vals, marker='o', label="Initial Trajectory")
plt.xlabel('X Position')
plt.ylabel('Y Position')
plt.title('Initial Trajectory')
plt.legend()
plt.grid(True)
plt.show()
```

```python

# Load the trajectory from the file
states = []
controls = []

with open(init_fname, "r") as f:
    for line in f:
        # Split the line into state (13 values) and control (6 values)
        data = list(map(float, line.strip().split()))
        state = data[:13]  # First 13 values are the state
        states.append(state)
        if len(data) > 13:
            control = data[13:]  # Remaining 6 values are the control
            controls.append(control)

# Extract X and Y positions from states
x_vals = [state[0] for state in states]  # X positions
y_vals = [state[1] for state in states]  # Y positions

# Plot the trajectory in the XY plane
plt.figure(figsize=(8, 6))
plt.plot(x_vals, y_vals, marker='o', label="Optimized Trajectory")
plt.xlabel('X Position')
plt.ylabel('Y Position')
plt.title('Optimized Trajectory in XY Plane')
plt.legend()
plt.grid(True)
plt.show()

# Optional: Plot controls if they exist
if controls:
    control_indices = range(len(controls))
    plt.figure(figsize=(8, 6))
    for i in range(6):
        control_vals = [control[i] for control in controls]
        plt.plot(control_indices, control_vals, label=f"u[{i}]")
    plt.xlabel('Time Step')
    plt.ylabel('Control Input')
    plt.title('Control Inputs over Time')
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
import matplotlib.pyplot as plt

for i in range(33):
    fname = f"../../../../granite_optim_trajectory_{i+1}.txt"
        
    # Load the trajectory from the file
    states = []
    controls = []

    with open(fname, "r") as f:
        for line in f:
            # Split the line into state (13 values) and control (6 values)
            data = list(map(float, line.strip().split()))
            state = data[:13]  # First 13 values are the state
            states.append(state)
            if len(data) > 13:
                control = data[13:]  # Remaining 6 values are the control
                controls.append(control)

    # Extract X, Y positions, linear velocities, quaternions, and angular velocities from states
    x_vals = [state[0] for state in states]  # X positions
    y_vals = [state[1] for state in states]  # Y positions
    lin_vel_x_vals = [state[3] for state in states]  # Linear velocities in X
    lin_vel_y_vals = [state[4] for state in states]  # Linear velocities in Y
    lin_vel_z_vals = [state[5] for state in states]  # Linear velocities in Z
    quaternions = [state[6:10] for state in states]  # Quaternion components [q_x, q_y, q_z, q_w]
    ang_vel_x_vals = [state[10] for state in states]  # Angular velocities in X
    ang_vel_y_vals = [state[11] for state in states]  # Angular velocities in Y
    ang_vel_z_vals = [state[12] for state in states]  # Angular velocities in Z

    # Extract linear control inputs (forces) and nonlinear control inputs (moments)
    forces_x_vals = [control[0] for control in controls]  # Forces in X
    forces_y_vals = [control[1] for control in controls]  # Forces in Y
    forces_z_vals = [control[2] for control in controls]  # Forces in Z
    moments_x_vals = [control[3] for control in controls]  # Moments in X
    moments_y_vals = [control[4] for control in controls]  # Moments in Y
    moments_z_vals = [control[5] for control in controls]  # Moments in Z

    # Indices for plotting
    t_indices = range(len(states))
    control_indices = range(len(controls))

    # Create a single figure with 6 subplots in a 2x3 layout
    fig, axs = plt.subplots(2, 3, figsize=(16, 12))

    # Plot 1: Position (X, Y)
    axs[0, 0].plot(x_vals, y_vals, marker='o', label="Optimized Trajectory")
    axs[0, 0].set_xlabel('X Position')
    axs[0, 0].set_ylabel('Y Position')
    axs[0, 0].set_xlim(-1, 1)
    axs[0, 0].set_ylim(1, -1)
    axs[0, 0].set_title(f'Position (case {i+1})')
    axs[0, 0].legend()
    axs[0, 0].grid(True)

    # Plot 2: Linear velocities (x, y, z)
    axs[0, 1].plot(t_indices, lin_vel_x_vals, label="v_x")
    axs[0, 1].plot(t_indices, lin_vel_y_vals, label="v_y")
    axs[0, 1].plot(t_indices, lin_vel_z_vals, label="v_z")
    axs[0, 1].set_xlabel('Time Step')
    axs[0, 1].set_ylabel('Linear Velocity')
    axs[0, 1].set_title(f'Linear Velocity (case {i+1})')
    axs[0, 1].legend()
    axs[0, 1].grid(True)

    # Plot 3: Forces (x, y, z)
    axs[0, 2].plot(control_indices, forces_x_vals, label="f_x")
    axs[0, 2].plot(control_indices, forces_y_vals, label="f_y")
    axs[0, 2].plot(control_indices, forces_z_vals, label="f_z")
    axs[0, 2].set_xlabel('Time Step')
    axs[0, 2].set_ylabel('Force')
    axs[0, 2].set_title(f'Forces (case {i+1})')
    axs[0, 2].legend()
    axs[0, 2].grid(True)

    # Plot 4: Quaternions (q_x, q_y, q_z, q_w)
    for k in range(4):
        q_vals = [q[k] for q in quaternions]
        axs[1, 0].plot(t_indices, q_vals, label=f"q[{k}]")
    axs[1, 0].set_xlabel('Time Step')
    axs[1, 0].set_ylabel('Quaternion Component')
    axs[1, 0].set_title(f'Quaternions (case {i+1})')
    axs[1, 0].legend()
    axs[1, 0].grid(True)

    # Plot 5: Angular velocities (x, y, z)
    axs[1, 1].plot(t_indices, ang_vel_x_vals, label="omega_x")
    axs[1, 1].plot(t_indices, ang_vel_y_vals, label="omega_y")
    axs[1, 1].plot(t_indices, ang_vel_z_vals, label="omega_z")
    axs[1, 1].set_xlabel('Time Step')
    axs[1, 1].set_ylabel('Angular Velocity')
    axs[1, 1].set_title(f'Angular Velocity (case {i+1})')
    axs[1, 1].legend()
    axs[1, 1].grid(True)

    # Plot 6: Moments (x, y, z)
    axs[1, 2].plot(control_indices, moments_x_vals, label="m_x")
    axs[1, 2].plot(control_indices, moments_y_vals, label="m_y")
    axs[1, 2].plot(control_indices, moments_z_vals, label="m_z")
    axs[1, 2].set_xlabel('Time Step')
    axs[1, 2].set_ylabel('Moment')
    axs[1, 2].set_title(f'Moments (case {i+1})')
    axs[1, 2].legend()
    axs[1, 2].grid(True)

    # Adjust layout for better spacing
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib
from mpl_toolkits.mplot3d import Axes3D  # Required for 3D plotting
%matplotlib widget

# Update file names or paths as necessary
fnames = [f"../../../../planner_scp_gusto_outputs/output_{i+1}.txt" for i in range(7)]

for i, fname in enumerate(fnames):
    states = []
    controls = []

    with open(fname, "r") as f:
        # First line: Solved status
        solved = f.readline().strip().split(":")[1].strip()
        print(f"Case {i+1}: Solved status - {solved}")

        # Next two lines: Initial and goal states
        x0 = list(map(float, f.readline().strip().split()))
        xg = list(map(float, f.readline().strip().split()))

        # Next line: Number of time steps (N)
        N = int(f.readline().strip())

        # Following N lines: State data
        for _ in range(N):
            state = list(map(float, f.readline().strip().split()))
            states.append(state)

        # Following N-1 lines: Control data
        for _ in range(N - 1):
            control = list(map(float, f.readline().strip().split()))
            controls.append(control)

    # Extract X, Y positions, linear velocities, quaternions, and angular velocities from states
    x_vals = [state[0] for state in states]
    y_vals = [state[1] for state in states]
    z_vals = [state[2] for state in states]
    lin_vel_x_vals = [state[3] for state in states]
    lin_vel_y_vals = [state[4] for state in states]
    lin_vel_z_vals = [state[5] for state in states]
    quaternions = [state[6:10] for state in states]  # [q_x, q_y, q_z, q_w]
    ang_vel_x_vals = [state[10] for state in states]
    ang_vel_y_vals = [state[11] for state in states]
    ang_vel_z_vals = [state[12] for state in states]

    # Extract forces and moments from controls
    if controls:  # Ensure controls are available
        forces_x_vals = [control[0] for control in controls]
        forces_y_vals = [control[1] for control in controls]
        forces_z_vals = [control[2] for control in controls]
        moments_x_vals = [control[3] for control in controls]
        moments_y_vals = [control[4] for control in controls]
        moments_z_vals = [control[5] for control in controls]

    # Indices for plotting
    t_indices = range(len(states))
    control_indices = range(len(controls))

    # Create a single figure with 6 subplots in a 2x3 layout
    fig, axs = plt.subplots(2, 3, figsize=(16, 12))

    # Plot 1: Position (X, Y, Z) in 3D
    axs[0, 0].remove()  # Remove 2D plot
    ax_3d = fig.add_subplot(2, 3, 1, projection='3d')
    ax_3d.plot(x_vals, y_vals, z_vals, marker='o', label="Trajectory")
    ax_3d.set_xlabel('X Position')
    ax_3d.set_ylabel('Y Position')
    ax_3d.set_zlabel('Z Position')
    ax_3d.set_title(f'Position (case {i+1})')
    ax_3d.legend()
    ax_3d.grid(True)

    # Plot 2: Linear velocities
    axs[0, 1].plot(t_indices, lin_vel_x_vals, label="v_x")
    axs[0, 1].plot(t_indices, lin_vel_y_vals, label="v_y")
    axs[0, 1].plot(t_indices, lin_vel_z_vals, label="v_z")
    axs[0, 1].set_xlabel('Time Step')
    axs[0, 1].set_ylabel('Linear Velocity')
    axs[0, 1].set_title(f'Linear Velocity (case {i+1})')
    axs[0, 1].legend()
    axs[0, 1].grid(True)

    # Plot 3: Forces
    if controls:
        axs[0, 2].plot(control_indices, forces_x_vals, label="f_x")
        axs[0, 2].plot(control_indices, forces_y_vals, label="f_y")
        axs[0, 2].plot(control_indices, forces_z_vals, label="f_z")
    axs[0, 2].set_xlabel('Time Step')
    axs[0, 2].set_ylabel('Force')
    axs[0, 2].set_title(f'Forces (case {i+1})')
    axs[0, 2].legend()
    axs[0, 2].grid(True)

    # Plot 4: Quaternions
    axs[1, 0].plot(t_indices, [q[0] for q in quaternions], label="q_x")
    axs[1, 0].plot(t_indices, [q[1] for q in quaternions], label="q_y")
    axs[1, 0].plot(t_indices, [q[2] for q in quaternions], label="q_z")
    axs[1, 0].plot(t_indices, [q[3] for q in quaternions], label="q_w")
    axs[1, 0].set_xlabel('Time Step')
    axs[1, 0].set_ylabel('Quaternion')
    axs[1, 0].set_title(f'Quaternions (case {i+1})')
    axs[1, 0].legend()
    axs[1, 0].grid(True)

    # Plot 5: Angular velocities
    axs[1, 1].plot(t_indices, ang_vel_x_vals, label="omega_x")
    axs[1, 1].plot(t_indices, ang_vel_y_vals, label="omega_y")
    axs[1, 1].plot(t_indices, ang_vel_z_vals, label="omega_z")
    axs[1, 1].set_xlabel('Time Step')
    axs[1, 1].set_ylabel('Angular Velocity')
    axs[1, 1].set_title(f'Angular Velocity (case {i+1})')
    axs[1, 1].legend()
    axs[1, 1].grid(True)

    # Plot 6: Moments
    if controls:
        axs[1, 2].plot(control_indices, moments_x_vals, label="m_x")
        axs[1, 2].plot(control_indices, moments_y_vals, label="m_y")
        axs[1, 2].plot(control_indices, moments_z_vals, label="m_z")
    axs[1, 2].set_xlabel('Time Step')
    axs[1, 2].set_ylabel('Moment')
    axs[1, 2].set_title(f'Moments (case {i+1})')
    axs[1, 2].legend()
    axs[1, 2].grid(True)

    # Adjust layout and display
    plt.tight_layout()
    plt.show()
```
